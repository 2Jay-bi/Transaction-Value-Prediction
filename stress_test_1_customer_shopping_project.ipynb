```
# Predictor variables
X = encoded_df.drop(
    columns=[
        "price",
        "invoice_no",
        "customer_id",
        "invoice_date"
    ],
    errors="ignore"
)

# Target variable
y = encoded_df["price"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```
non_numeric = X.select_dtypes(exclude=["number", "bool"]).columns

print("Remaining non-numeric columns:")
print(non_numeric.tolist())
```

```
bool_columns = X.select_dtypes(include="bool").columns

X[bool_columns] = X[bool_columns].astype(int)

print(X.dtypes.value_counts())
```

```
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training set:", X_train.shape)
print("Testing set:", X_test.shape)
```

```
from sklearn.ensemble import RandomForestRegressor

rf_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train, y_train)

print("Random Forest Model Trained Successfully!")
```

```
import numpy as np
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

y_pred = rf_model.predict(X_test)

mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("Random Forest Model Performance")
print("--------------------------------")
print(f"MAE:  {mae:.2f}")
print(f"RMSE: {rmse:.2f}")
print(f"R²:   {r2:.4f}")
```

```
import matplotlib.pyplot as plt

# Create Predicted vs Actual plot
plt.figure(figsize=(8,6))
plt.scatter(y_test, y_pred, alpha=0.5)

# Reference line (perfect predictions)
plt.plot(
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()],
    color='red',
    linewidth=2
)

plt.title("Predicted vs Actual Purchase Values")
plt.xlabel("Actual Purchase Value")
plt.ylabel("Predicted Purchase Value")

plt.grid(True)
plt.show()
```

```
import pandas as pd
import matplotlib.pyplot as plt

# Calculate feature importance
feature_importance = pd.DataFrame({
    'Feature': X.columns,
    'Importance': rf_model.feature_importances_
})

# Select the top 10 features
feature_importance = feature_importance.sort_values(
    by='Importance',
    ascending=False
).head(10)

# Plot
plt.figure(figsize=(10,6))

plt.barh(
    feature_importance['Feature'],
    feature_importance['Importance']
)

plt.xlabel("Importance Score")
plt.title("Top 10 Feature Importance")

plt.gca().invert_yaxis()

plt.show()
```
